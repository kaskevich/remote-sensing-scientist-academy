## Lesson 5 — Repetition, loops and vectorised thinking

Define the batch before running it: analysis population, field meaning, inclusion and missing-value policy, repeated operation, outputs and independent hand check.

```python
plots_1_to_6 = [
    {"SampleID": "SALS1", "Sp_richness": 7},
    {"SampleID": "SALS2", "Sp_richness": 6},
    {"SampleID": "SALS3", "Sp_richness": 5},
    {"SampleID": "SALS4", "Sp_richness": 7},
    {"SampleID": "SALS5", "Sp_richness": 7},
    {"SampleID": "SALS6", "Sp_richness": 3},
]
```

```python
review_threshold = 6  # Instructional only
records_received = 0
values_used = 0
missing_values = 0
richness_total = 0
review_ids = []

for plot in plots_1_to_6:
    records_received = records_received + 1
    richness = plot["Sp_richness"]
    if richness is None:
        missing_values = missing_values + 1
    else:
        values_used = values_used + 1
        richness_total = richness_total + richness
        if richness < review_threshold:
            review_ids.append(plot["SampleID"])

mean_richness = richness_total / values_used if values_used > 0 else None
print(records_received, values_used, missing_values, richness_total, mean_richness, review_ids)
```

## Lesson 6 — Functions, errors and debugging

Document the function's purpose, accepted input representations, returned output, source-mutation policy, known limitation and scientific assumption before defining it.

```python
def classify_biomass(value):
    """Classify one AGB value using Lesson 6 value types."""
    if value is None:
        return "missing"
    if isinstance(value, bool):
        return "invalid Boolean"
    if not isinstance(value, (int, float)):
        return "invalid type"
    if value < 0:
        return "invalid negative"
    return "recorded"
```

## Lesson 9 — Missing values, types and data quality

Profile completeness, conformance, consistency and identity separately. Keep every rule, denominator, provenance, flag, reversible action and unresolved risk beside the source evidence.

```python
def classify_biomass_pandas(value):
    """Classify one scalar AGB value from a pandas workflow."""
    if isinstance(value, (bool, np.bool_)):
        return "invalid Boolean"
    if pd.isna(value):
        return "missing"
    if not isinstance(value, (int, float, np.integer, np.floating)):
        return "invalid type"
    if value < 0:
        return "invalid negative"
    return "recorded"
```

```python
assert classify_biomass_pandas(None) == "missing"
assert classify_biomass_pandas(np.nan) == "missing"
assert classify_biomass_pandas(311.33) == "recorded"
assert classify_biomass_pandas(-1) == "invalid negative"
assert classify_biomass_pandas("311.33") == "invalid type"
assert classify_biomass_pandas(True) == "invalid Boolean"
```

```python
agb_coverage = (
    meadows.assign(AGB_available=meadows["AGB"].notna())
    .groupby(["site", "plantcommunity"], observed=True)["AGB_available"]
    .agg(available_n="sum", total_n="size")
)
agb_coverage["available_pct"] = 100 * agb_coverage["available_n"] / agb_coverage["total_n"]
print(agb_coverage)
```
